```python
text = "You can use the tool below to understand how a piece of text might be tokenized by a language model, and the total count of tokens in that piece of text."
print(text)
```

```python
tokens = text.split()
print(tokens)
```

```python
vocab = {}
current_id = 0

for token in tokens:
  if token not in vocab:
    vocab[token] = current_id
    current_id += 1

print(vocab)
```

```python
encoded = [vocab[token] for token in tokens]
print(encoded)
```

```python
reversed_vocab = {id: token for token, id in vocab.items()}
decoded = " ".join(reversed_vocab[id] for id in encoded)
print(decoded)
```

```python
# Embedding
# "hello" => [0.41, 0.22, -0.77, 0.56 ......]
# "hi" => [0.43, 0.72, 0.77, 0.88 ......]
```

```python
import torch
import torch.nn as nn

embedding_dim = 0
vocab_size = len(vocab)

embedding_layer = nn.Embedding(vocab_size, embedding_dim)

input_ids = torch.tensor([encoded])
embeddings = embedding_layer(input_ids)

print(embeddings)
print(embeddings.shape)
```

```python
from transformers import AutoTokenizer

model_id = "TinyLlama/TinyLlama-1.1b-Chat-v1.0"
tokenizer = AutoTokenizer.from_pretrained(model_id)
```

```python
print('Vocab size:', tokenizer.vocab_size)
print('Special tokens:', tokenizer.special_tokens_map)
```

```python
text = "Hello how are you doing?"
print("text:", text)
```

```python
tokens = tokenizer.tokenize(text)
print("tokens:", tokens)
```

```python
ids = tokenizer.convert_tokens_to_ids(tokens)
print("ids:", ids)
```

```python
encoded = tokenizer(text)
print("encoded:", encoded)
```

```python
decoded_text = tokenizer.decode(encoded['input_ids'])
print("decoded_text:", decoded_text)
```

```python
texts = [
    "Hello, how are you?",
    "Today we are learning tokenization.",
    "This is a sample text for tokenization."
]

batch = tokenizer(texts, padding=True, return_tensors='pt')
print(batch.keys())

print("Input IDs shape:", batch['input_ids'].shape)
print("Attention Mask shape:", batch['attention_mask'].shape)
```
