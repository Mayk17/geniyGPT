```python
text = "You can use the tool below to understand how a piece of text might be tokenized by a language model, and the total count of tokens in that piece of text."
print(text)
```

```python
tokens = text.split()
print(tokens)
```

```python
vocab = {}
current_id = 0

for token in tokens:
  if token not in vocab:
    vocab[token] = current_id
    current_id += 1

print(vocab)
```

```python
encoded = [vocab[token] for token in tokens]
print(encoded)
```

```python
reversed_vocab = {id: token for token, id in vocab.items()}
decoded = " ".join(reversed_vocab[id] for id in encoded)
print(decoded)
```

```python
# Embedding
# "hello" => [0.41, 0.22, -0.77, 0.56 ......]
# "hi" => [0.43, 0.72, 0.77, 0.88 ......]
```

```python
import torch
import torch.nn as nn

embedding_dim = 0
vocab_size = len(vocab)

embedding_layer = nn.Embedding(vocab_size, embedding_dim)

input_ids = torch.tensor([encoded])
embeddings = embedding_layer(input_ids)

print(embeddings)
print(embeddings.shape)
```
